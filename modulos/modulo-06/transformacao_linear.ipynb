## Caso de Uso: Rotação de Imagens

#### Rotacionar uma imagem em _quartos de volta_ (90°, 180° e 270°)

Definir caminho de leitura da imagem

```python
IMG_PATH = "../../assets/lena.jpg"
```

Carregar e tratar a imagem

```python
# Ler os bytes da imagem:
img_bytes = tf.io.read_file(filename=IMG_PATH)


# Decodificar para o "rank 3 tensor" (dimensões H, W, C):
img = tf.image.decode_jpeg(img_bytes, channels=3)


# Devido a imagem da Lena estar em preto e branco,
# é necessário adicionar mais um canal de dimensão:
if tf.rank(img) == 2:
    img = img[..., tf.newaxis]
```

Plotar a imagem original

```python
plt.imshow(img)
```

Rotacionar em 90°

```python
img_90 = tf.image.rot90(img, k=1)
```

Plotar imagem rotacionada

```python
plt.imshow(img_90)
```

Rotacionar em 180°

```python
img_180 = tf.image.rot90(img, k=2)
```

Plotar imagem rotacionada

```python
plt.imshow(img_180)
```

Rotacionar em 270°

```python
img_270 = tf.image.rot90(img, k=3)
```

Plotar imagem rotacionada

```python
plt.imshow(img_270)
```
