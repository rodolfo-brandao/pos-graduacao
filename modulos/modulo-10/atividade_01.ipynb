# Predição de Dados e Avaliação de Modelos

Utilizando os datasets _Iris_ e _Titanic_

## Imports

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB  # Naive Bayes Gaussiano
from sklearn.linear_model import LogisticRegression  # Regressão Logística
from sklearn.datasets import load_iris  # Dataset Iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
```

## 1. Dataset Iris

### Carregar Dados

```python
iris_data = load_iris()
```

### Definir _Atributos_ e _Rótulos_

**X** e **y** para treino e teste

```python
(
    X_train,
    X_test,
    y_train,
    y_test
) = train_test_split(iris_data.data, iris_data.target, test_size=0.3, random_state=1)  # type: ignore (Pylance)
```

### Modelo: Naive Bayes Gaussiano

```python
iris_model = GaussianNB()
iris_model.fit(X_train, y_train)
```

#### Predição

```python
y_pred = iris_model.predict(X_test)
print(y_pred)
```

#### Avaliação

```python
confusion_matrix(y_test, y_pred)  # Comparar previsões com rótulos reais

print(confusion_matrix(y_test, y_pred))
print(f"Acurácia: {accuracy_score(y_test, y_pred):.1%}")
```

## 2. Dataset Titanic

### Carregar Dados

```python
titanic_df = pd.read_csv(
    filepath_or_buffer="https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
)
```

### Pré-processamento dos Dados

1. Selecionar apenas colunas relevantes

```python
titanic_copy_df = titanic_df[['Pclass', 'Sex', 'Age', 'Fare', 'Survived']].copy()
```

2. Preencher valores faltantes

```python
titanic_copy_df['Age'] = titanic_copy_df['Age'].fillna(titanic_copy_df['Age'].mean())
titanic_copy_df['Fare'] = titanic_copy_df['Fare'].fillna(titanic_copy_df['Fare'].mean()) 
```

3. Converter coluna `Sex` para numérico

```python
titanic_copy_df['Sex'] = titanic_copy_df['Sex'].map({'male': 0, 'female': 1})
```

### Definir _Atributos_ e _Rótulos_

**X** e **y** para treino e teste

```python
X = titanic_copy_df.drop('Survived', axis=1)
y = titanic_copy_df['Survived']

(
    X_train,
    X_test,
    y_train,
    y_test
) = train_test_split(X, y, test_size=0.3, random_state=1)
```

### Modelo: Naive Bayes Gaussiano

```python
titanic_model = GaussianNB()
titanic_model.fit(X_train, y_train)
```

#### Predição

```python
y_pred = titanic_model.predict(X_test)
print(y_pred)
```

#### Avaliação

```python
print(confusion_matrix(y_test, y_pred))
print(f"Acurácia: {accuracy_score(y_test, y_pred):.1%}")
```

### Modelo: Regressão Logística

```python
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
```

#### Predição

```python
lr_pred = lr_model.predict(X_test)
print(lr_pred)
```

#### Avaliação

```python
print(confusion_matrix(y_test, lr_pred))
print(f"Acurácia: {accuracy_score(y_test, lr_pred):.1%}")
```
