# Principal Component Analysis (PCA)

Referências
- [The Iris Dataset](https://scikit-learn.org/1.4/auto_examples/datasets/plot_iris_dataset.html)
- [scikit-learn PCA library](https://scikit-learn.org/1.4/modules/generated/sklearn.decomposition.PCA.html#sklearn.decomposition.PCA)

## Imports

```python
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

## Carregar Dataset *Iris*

```python
iris = datasets.load_iris()
```

## Definir *features* e *targets*

Atribuir em variáveis `X` e `y` as *features* e *targets*, respectivamente

- Features (length/width):
    - Sepal
    - Petal
- Targets:
    - 0 = Setosa
    - 1 = Versicolor
    - 2 = Virginica

```python
X = iris.data
y = iris.target

iris.feature_names
```

## Normalizar os Dados

Necessário, pois o PCA é sensível à escala dos dados

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

Reduzir de 4D para 2D, para visualização

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X=X_scaled)
```

## Visualizar Dados Transformados

```python
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', edgecolor='k', s=80)
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.title('Análise do dataset Íris')
plt.show()
```

## Checar Variância dos Dados

```python
print("Taxa de variância explicada:", pca.explained_variance_ratio_)
print("Total de variância explicada:", sum(pca.explained_variance_ratio_))
```
