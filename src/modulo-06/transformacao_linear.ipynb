### Imports

```python
import matplotlib.pyplot as plt
import tensorflow as tf
```

### Rotacionar Imagem

Declarar constantes auxiliares

```python
IMG_READ_PATH = "assets/lena.jpg"
IMG_90_SAVE_PATH = "assets/lena_90.jpg"
IMG_180_SAVE_PATH = "assets/lena_180.jpg"
```

Carregar e tratar a imagem

```python
# Ler os bytes da imagem:
img_bytes = tf.io.read_file(filename=IMG_READ_PATH)


# Decodificar para o "rank 3 tensor" (dimensões H, W, C):
img = tf.image.decode_jpeg(img_bytes, channels=3)


# Devido a imagem da Lena estar em preto e branco,
# é necessário adicionar mais um canal de dimensão:
if tf.rank(img) == 2:
    img = img[..., tf.newaxis]
```

Plotar a imagem original

```python
plt.imshow(img)
```

Rotacionar em 90°

```python
img_90 = tf.image.rot90(img, k=1)
```

Plotar imagem rotacionada

```python
plt.imshow(img_90)
```

Rotacionar em 180°

```python
img_180 = tf.image.rot90(img, k=2)
```

Plotar imagem rotacionada

```python
plt.imshow(img_180)
```
