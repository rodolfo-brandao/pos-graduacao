## Imports

```python
import tensorflow as tf
```

## Rotacionar Imagem

```python
# Declarar os caminhos de leitura e escrita das imagens em uma constante para melhor reusabilidade:
IMG_READ_PATH = "assets/lena.jpg"
IMG_90_SAVE_PATH = "assets/lena_90.jpg"
IMG_180_SAVE_PATH = "assets/lena_180.jpg"
```

Usando o TensorFlow...

```python
# Ler os bytes da imagem:
img_bytes = tf.io.read_file(filename=IMG_READ_PATH)

# Decodificar para o "rank 3 tensor" (dimensões H, W, C):
img = tf.image.decode_jpeg(img_bytes, channels=3)


# Devido a imagem da Lena ser em preto e branco,
# é necessário adicionar mais um canal de dimensão:
if tf.rank(img) == 2:
    img = img[..., tf.newaxis]
```

Rotacionar em 90°

```python
# Rotacionar a imagem em 90° (k=1)
img_90 = tf.image.rot90(img, k=1)


# Salvar de volta a imagem rotacionada no disco:
tf.io.write_file(
    filename=IMG_90_SAVE_PATH,
    contents=tf.io.encode_jpeg(img_90)
)
```

Rotacionar em 180°

```python
# Rotacionar a imagem em 180° (k=2)
img_180 = tf.image.rot90(img, k=2)


# Salvar de volta a imagem rotacionada no disco:
tf.io.write_file(
    filename=IMG_180_SAVE_PATH,
    contents=tf.io.encode_jpeg(img_180)
)
```
